# Carousel UI assumption

![Carousel UI](../img/carousel.png)

This notebook studies the Expedia ranking data under a **carousel behavior** assumption.

Here, the user is assumed to browse hotels through **small visible horizontal groups** instead of one long list. In this kind of UI, raw position is not enough, because it mixes together how deep the user had to go and where the hotel appeared inside the visible strip.

For this notebook, the key function is `get_carousel_dataset`.  
It is the right transformation because it decomposes each ranking position into two UI-driven coordinates: `carousel_row` and `slot`. That gives the dataset the structure needed to represent a carousel, where visibility depends both on depth and on within-row placement.

So the purpose of this notebook is to analyze the ranking as a **grouped horizontal browsing interface** rather than as a simple top-to-bottom list.
