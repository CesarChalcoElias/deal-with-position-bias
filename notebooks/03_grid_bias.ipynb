# Grid / PLP UI assumption

![Grid UI](../img/grid.png)

This notebook studies the Expedia ranking data under a **PLP-style behavior** assumption.

PLP stands for **product listing page**. In this context, think of a classic old Expedia results page where several hotels are visible immediately, while the rest require scrolling.

For this notebook, the key function is `get_grid_dataset`.  
It is the right transformation because it adds a simple visibility signal, `above_fold`, based on whether the hotel position is within the fold threshold. That matches the main behavioral distinction in a classic listing page: some results are visible without scrolling, and others are not.

So the purpose of this notebook is to analyze the ranking as a **traditional search results page**, where above-the-fold exposure is one of the main drivers of attention.
